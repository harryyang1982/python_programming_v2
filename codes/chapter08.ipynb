#03. 리스트 컴프리헨션

```python
result = []
for i in range(10):
    result.append(i)

result
```

```python
result = [i for i in range(10)]
result
```

```python
result = []
for i in range(10):
    if i % 2 ==0:
        result.append(i)

result
```

```python
result = [i for i in range(10) if i % 2 == 0]
result
```

```python
result = [i if i % 2 == 0 else 10 for i in range(10)]
result
```

```python
word_1 = "Hello"
word_2 = "World"
result = [i + j for i in word_1 for j in word_2]
result
```

```python
case_1 = ["A", "B", "C"]
case_2 = ["D", "E", "A"]
result = [i + j for i in case_1 for j in case_2 if not(i==j)]
result
```

```python
words = 'The quick brown fox jumps over the lazy dog'.split()
print(words)
```

```python
stuff = [[w.upper(), w.lower(), len(w)] for w in words]
```

```python
stuff
```

```python
for i in stuff:
    print(i)
```

```python
case_1 = ["A", "B", "C"]
case_2 = ["D", "E", "A"]
result = [i + j for i in case_1 for j in case_2]
result
```

```python
result = [[i + j for i in case_1] for j in case_2]
result
```

# 04. 다양한 방식의 리스트값 출력

```python
for i,v in enumerate(['tic', 'tac', 'toe']):
    print(i, v)
```

```python
{i:j for i, j in enumerate('TEAMLAB is an academic institute located in South Korea.'.split())}
```

```python
alist = ['a1', 'a2', 'a3']
blist = ['b1', 'b2', 'b3']
for a, b in zip(alist, blist):
    print(a, b)
```

```python
a, b, c = zip((1, 2, 3), (10, 20, 30), (100, 200, 300))
print(a, b, c)
```

```python
[sum(x) for x in zip((1, 2, 3), (10, 20, 30), (100, 200, 300))]
```

```python
alist = ['a1', 'a2', 'a3']
blist = ['b1', 'b2', 'b3']
```

```python
for i, (a, b) in enumerate(zip(alist, blist)):
    print(i, a, b)
```

```python
mylist = ['pen', 'pencil', 'sharp']
result = list(enumerate(mylist))
print(result)
```

```python
kor_score = [70, 79, 80, 90, 80]
math_score = [42, 80, 30, 50, 90]
eng_score = [53, 77, 50, 70, 55]
midterm_score = [kor_score, math_score, eng_score]
print("score:", midterm_score[2][1])
```

```python
user_dict = {}
user_list = ['students', 'superuser', 'professor', 'employee']
user_dict = {value_2:value_1 for value_1, value_2 in enumerate(user_list)}
print(user_dict)
```

```python
alphabet = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h']
nums = [i for i in range(20)]
answer = [alpha+str(num) for alpha in alphabet for num in nums if num%2 == 0]
print(len(answer))
```
