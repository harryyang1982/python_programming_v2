# 02. 맵리듀스

```python
ex = [1, 2, 3, 4, 5]
f = lambda x: x ** 2
print(list(map(f, ex)))
```

```python
ex = [1, 2, 3, 4, 5]
f = lambda x: x ** 2
for value in map(f, ex):
    print(value)
```

```python
ex = [1, 2, 3, 4, 5]
[x ** 2 for x in ex]
```

```python
ex = [1, 2, 3, 4, 5]
f = lambda x, y: x+ y
list(map(f, ex, ex))
```

```python
[x + y for x, y in zip(ex, ex)]
```

```python
list(map(lambda x: x ** 2 if x % 2 == 0 else x, ex))
```

```python
[x ** 2 if x % 2 == 0 else x for x in ex]
```

```python
from functools import reduce
```

```python
print(reduce(lambda x, y: x+y, [1, 2, 3, 4, 5]))
```

```python
x = 0
for y in [1, 2, 3, 4, 5]:
    x += y

print(x)
```

# 03. 별표의 활용

```python
def asterisk_test(a, *args):
    print(a, args)
    print(type(args))
    
asterisk_test(1, 2, 3, 4, 5, 6)
```

```python
def asterisk_test(a, **kargs):
    print(a, kargs)
    print(type(kargs))

asterisk_test(1, b=2, c=3, d=4, e=5, f=6)
```

```python
def asterisk_test(a, args):
    print(a, *args)
    print(type(args))
    
asterisk_test(1, (2, 3, 4, 5, 6))
```

```python
def asterisk_test(a, *args):
    print(a, args)
    print(type(args))
    
asterisk_test(1, *(2, 3, 4, 5, 6))
```

```python
a, b, c = ([1, 2], [3, 4], [5, 6])
print(a, b, c)
```

```python
data = ([1, 2], [3, 4], [5, 6])
print(*data)
```

```python
for data in zip(*[[1, 2], [3, 4], [5, 6]]):
    print(data)
    print(type(data))
```

```python
def asterisk_test(a, b, c, d,):
    print(a, b, c, d)

data = {'b':1, 'c':2, 'd':3}
asterisk_test(10, **data)
```

```python
u = [2, 2]
v = [2, 3]
z = [3, 5]
result = []
```

```python
for i in range(len(u)):
    result.append(u[i] + v[i] + z[i])

print(result)
```

```python
result = [sum(t) for t in zip(u, v, z)]
print(result)
```

```python
[t for t in zip(u, v, z)]
```

```python
def vector_addition(*args):
    return [sum(t) for t in zip(*args)]

vector_addition(u, v, z)
```

```python
row_vectors = [[2, 2], [2, 3], [3, 5]]
vector_addition(*row_vectors)
```
