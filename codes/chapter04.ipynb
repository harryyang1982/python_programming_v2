# 04. Lab: 구구단 계산기

```python
gugu = int(input("구구단 몇 단을 계산할까?\n"))
print("구구단", gugu, "단을 계산한다.")
for i in range(1, 10):
    result = gugu * i
    print(gugu, "X", i, "=", result)
```

```python
gugu = int(input("구구단 몇 단을 계산할까?\n"))
if gugu not in range(1, 10):
    print("범위 오류")
else:
    print("구구단", gugu, "단을 계산한다.")
    for i in range(1, 10):
        result = gugu * i
        print(gugu, "X", i, "=", result)
```

# 06. Lab: 숫자 찾기 게임

```python
import random
game_number = int(input("숫자를 맞혀 보세요. (1 ~ 100)\n"))
answer = random.randint(1, 101)
while (game_number is not answer):
    if game_number > answer:
        print("숫자가 너무 큽니다")
        game_number = int(input("숫자를 맞혀 보세요. (1 ~ 100)\n"))
    elif game_number < answer:
        print("숫자가 너무 작습니다")
        game_number = int(input("숫자를 맞혀 보세요. (1 ~ 100)\n"))
else:
    print("정답입니다. 입력한 숫자는", answer, "입니다.")

```

# 07. Lab: 연속적인 구구단 계산기

```python
gugu = 1
while (gugu != 0):
    gugu = int(input("구구단 몇 단을 계산할까요(1~9)?\n"))
    if gugu == 0:
        break
    elif not(1 <= gugu <= 9):
        print("잘못 입력했습니다. 다시 하세요")
        continue
    else:
        print("구구단", str(gugu), "단을 계산합니다.")
        for i in range(1, 10):
            print(str(gugu), "X ", str(i) + "= " + str(gugu*i))
print("구구단 게임을 종료합니다.")
```
