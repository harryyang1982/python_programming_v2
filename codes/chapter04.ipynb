# 02. LAB: 어떤 종류의 학생인지 맞히기

```python
year = int(input("당신이 태어난 연도를 입력하세요: "))

age = 2026 - year + 1
print(age)

if age > 26:
    syear = '학생이 아닙니다'
elif age >= 20:
    syear = '대학생'
elif age >= 17:
    syear = '고등학생'
elif age >= 14:
    syear = '중학생'
elif age >= 8:
    syear = '초등학생'
else:
    syear = '학생이 아닙니다'

print(syear)
```
